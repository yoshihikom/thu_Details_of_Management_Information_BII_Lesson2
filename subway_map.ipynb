5.5.2 ランドマークの位置情報を取得する

```python
#Geocoder を用いてカンザスシティ（USのほぼ真ん中）の場所を取得する
import geopy
from geopy.geocoders import Nominatim

#geocoder - from location address to yx coordinates
geolocator = Nominatim(timeout=10, user_agent = "y.murakawa@thu.ac.jp")
location = geolocator.geocode('Kansas City')
print(location)
address=[location.latitude, location.longitude]
print(address)
```

US_subway_data.csvからデータを読み込む

```python
import pandas as pd
import io

#データを読み込む
input_data = pd.read_csv("US_subway_data.csv")

input_data.head(10)
```

```python
import folium
import pandas as pd

#利用するデータセットを決める
#緯度・経度・住所を取得する
places = pd.concat([input_data.latitude, input_data.longitude, input_data.street_address], axis=1)

#地図の表示箇所を決める
#今回はKansas Cityの位置情報を用いる
map = folium.Map(location=address, zoom_start=6)

#緯度・経度と場所情報をプロットする
for i, r in places.iterrows():
    folium.Marker(location=[r['latitude'], r['longitude']], popup=places['street_address']).add_to(map)

#地図をhtml ファイルとして保存する
map.save("map_sabway.html")
```

```python
import folium
#ヒートマップパッケージをインポートする
from folium.plugins import HeatMap

#緯度経度を指定する
places2 = pd.concat([input_data.latitude, input_data.longitude], axis=1)
map2 = folium.Map(location=address, zoom_start=6)

#緯度・経度と場所情報をプロットする
HeatMap(places2, radius=6, blur=5).add_to(map2)

#地図をhtml ファイルとして保存する
map2.save("map_subway_heatmap_airbnb.html", color="gray")
```
